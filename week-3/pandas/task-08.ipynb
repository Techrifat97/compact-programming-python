```python
import pandas as pd


dict1 = {
    "ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
}

dict2 = {
    "ID": [1, 2, 3],
    "Score": [85, 90, 78]
}


df1 = pd.DataFrame(dict1)
df2 = pd.DataFrame(dict2)


print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)


# Merge the two DataFrames
merged_df = pd.merge(df1, df2, on="ID")

print("\nMerged DataFrame:")
print(merged_df)


# Append Score as a new column to df1
df1["Score"] = df2["Score"]

print("\nDataFrame 1 after adding Score:")
print(df1)
```
