```python
import pandas as pd

df = pd.read_csv("../data/Cars93_missing.csv")

print("Missing Horsepower values before:")
print(df["Horsepower"].isnull().sum())

average = df["Horsepower"].mean()

df["Horsepower"] = df["Horsepower"].fillna(average)

print("\nAverage Horsepower:", average)

print("\nMissing Horsepower values after:")
print(df["Horsepower"].isnull().sum())
```
