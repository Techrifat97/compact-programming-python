```python
import pandas as pd

df = pd.read_csv("../data/Cars93_missing.csv")


def swap_columns(dataframe, column1, column2):
    columns = list(dataframe.columns)

    index1 = columns.index(column1)
    index2 = columns.index(column2)

    columns[index1], columns[index2] = columns[index2], columns[index1]

    return dataframe[columns]


# Exchange Manufacturer and Model
df = swap_columns(df, "Manufacturer", "Model")

print("After exchanging columns:")
print(df.head())


# Sort columns alphabetically
df = df.reindex(sorted(df.columns), axis=1)

print("\nColumns sorted by name:")
print(df.head())
```
