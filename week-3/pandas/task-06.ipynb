```python
import pandas as pd

df = pd.read_csv("../data/Cars93_missing.csv")

lower_limit = df["Price"].quantile(0.05)
upper_limit = df["Price"].quantile(0.95)

print("Lower 5% limit:", lower_limit)
print("Upper 5% limit:", upper_limit)

trimmed_df = df[
    (df["Price"] >= lower_limit) &
    (df["Price"] <= upper_limit)
]

print("\nOriginal rows:", len(df))
print("Rows after removing upper and lower 5%:", len(trimmed_df))

print(trimmed_df)
```
